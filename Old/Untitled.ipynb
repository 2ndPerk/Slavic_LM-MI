```python
from transformers import  AutoTokenizer

def see_tokens(tokenizer, d):
    t = tokenizer.tokenize(d)
    print(t)
    print(tokenizer.convert_tokens_to_ids(t))

tokenizer = AutoTokenizer.from_pretrained("spital/gpt2-small-czech-cs")
see_tokens(tokenizer, "dobrý deň")
see_tokens(tokenizer, "dobrý den")
```
