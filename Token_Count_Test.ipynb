```python
def count_tokens(tokenizer, data):
    r = 0
    for d in data:
        for s in d:
            for l in s:
                r += len(tokenizer.tokenize(l))
    return r
```

```python
def token_test(tokenizer):
    langs = ["Polish", "Czech", "Slovak", "Croatian", "Slovene"]
    results = []

    for lang in tqdm(langs):
        word_list = SMI_Methods.prep_words(lang)
        data = SMI_Methods.prep_data(lang, word_list)
        
        print(lang, count_tokens(tokenizer, data))
```
