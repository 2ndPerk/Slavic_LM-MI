```python
import torch
from transformers import  CamembertForMaskedLM, CamembertTokenizer, CamembertModel,RobertaTokenizer,RobertaModel,RobertaForMaskedLM, pipeline
from transformers import BertForMaskedLM, BertTokenizer
import math
import numpy as np
import json
import random
import pandas as pd
```

```python
#Create Masked Latin Data
source = "PolishCorpus.txt"
file = open(source, "r")
lines = file.readlines()

data = []

for l in range(5):
    line = lines[l]
    s = line.split(" ")
    s[-1] = s[-1].replace("\n", "")

    rand = random.randint(2,len(s)-2)

    word = s[rand]
    s[rand] = "{}"
    text = " ".join(s)

    data.append([text,word])
```

```python
def score_model(model, tokenizer, data):
    nlp = pipeline("fill-mask", model = model,tokenizer = tokenizer, top_k = 10)
    for i in range(5):
        opt = nlp(data[i][0].format(nlp.tokenizer.mask_token))
        print(data[i][0])
        print("Correct: ",data[i][1])
        for r in opt:
            print(r['token_str'])
        print()
    return
```

```python
#Polish
model = BertForMaskedLM.from_pretrained("dkleczek/bert-base-polish-uncased-v1")
tokenizer = BertTokenizer.from_pretrained("dkleczek/bert-base-polish-uncased-v1")
scoredModel = score_model(model, tokenizer, data)
```

```python
#Czech
tokenizer = RobertaTokenizer.from_pretrained('ufal/robeczech-base')
model = RobertaForMaskedLM.from_pretrained('ufal/robeczech-base')
scoredModel = score_model(model, tokenizer, data)
```

```python
#Slovak
tokenizer = RobertaTokenizer.from_pretrained('gerulata/slovakbert')
model = RobertaForMaskedLM.from_pretrained('gerulata/slovakbert')
scoredModel = score_model(model, tokenizer, data)
```
